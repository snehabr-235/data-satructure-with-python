```python

age_list = [24, 25, 26, 27, 28]

name_list = ["Anu", "Meera", "Rahul", "Arjun", "Diya"]

print("Age List:", age_list)
print("Name List:", name_list)
```

```python
name_list.append("Yazhini")
print(name_list)

age_list.insert(2, 30)
print(age_list)

name_list.remove("Yazhini")
print(name_list)

age_list.pop()
print(age_list)

age_list.extend([29, 30, 26])
print(age_list)

age_list.sort(reverse=True)
print(age_list)

print(max(age_list))
print(min(age_list))
print(sum(age_list))
```

```python
print(name_list[0])

print(name_list[-1])

print(name_list[2:5])

print(name_list[::-1])
```

```python
# a. Create a set
my_set = {'a', 'e', 'i', 'o', 'u', 'a', 'a', 'i'}

print("my_set:", my_set)
print("Explanation: Duplicate values are automatically removed because sets contain only unique elements.")

# b. Attempt to change a value
try:
    my_set[4] = 's'
except TypeError:
    print("Error: Sets are unordered and do not support indexing or changing elements.")

# c. Create two sets
set1 = {1, 3, 5, 7, 9}
set2 = {2, 3, 5, 8, 10}

# d. Union and intersection
print("Union:", set1.union(set2))
print("Intersection:", set1.intersection(set2))
```
